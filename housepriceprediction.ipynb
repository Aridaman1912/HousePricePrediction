```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
categorical_pipeline=Pipeline([
    ("imputer",SimpleImputer(strategy="most_frequent")),
    ("encoder",OneHotEncoder(handle_unknown="ignore"))
])

numerical_pipeline=Pipeline([
    ("imputer",SimpleImputer(strategy="median"))
])

preprocessor_pipeline=ColumnTransformer([
    ("num",numerical_pipeline,numerical_cols),
    ("cat",categorical_pipeline,categorical_cols)
])
```

```python
from sklearn.linear_model import LinearRegression

lr_pipeline = Pipeline([
    ("preprocessor", preprocessor_pipeline),
    ("model", LinearRegression())
])
```

```python
lr_pipeline.fit(X_train, y_train_log)
```

```python
y_pred_pipeline=lr_pipeline.predict(X_test)
```

```python
rmsle_pipeline=np.sqrt(
    np.mean((y_test_log-y_pred_pipeline)**2)
)
print("pipeline lr rmsle:",rmsle_pipeline)
```

```python
from sklearn.model_selection import cross_val_score

cv_scores_pipeline_lr = cross_val_score(
    lr_pipeline,
    X_train,
    y_train_log,
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

cv_rmsle_pipeline_lr = -cv_scores_pipeline_lr

print("Pipeline LR Fold RMSLE:", cv_rmsle_pipeline_lr)
print("Pipeline LR Mean RMSLE:", cv_rmsle_pipeline_lr.mean())
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb_pipeline = Pipeline([
    ("preprocessor", preprocessor_pipeline),
    ("model", GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    ))
])
```

```python
gb_pipeline.fit(X_train,y_train_log)
```

```python
y_pred_gb_pipeline=gb_pipeline.predict(X_test)
```

```python
rmsle_gb_pipeline = np.sqrt(
    np.mean((y_test_log - y_pred_gb_pipeline) ** 2)
)

print("Pipeline Gradient Boosting RMSLE:", rmsle_gb_pipeline)
```

```python
cv_scores_gb_pipeline = cross_val_score(
    gb_pipeline,
    X_train,
    y_train_log,
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

cv_rmsle_gb_pipeline = -cv_scores_gb_pipeline

print("Pipeline GB Fold RMSLE:", cv_rmsle_gb_pipeline)
print("Pipeline GB Mean RMSLE:", cv_rmsle_gb_pipeline.mean())
```

```python
from sklearn.model_selection import GridSearchCV

param_grid_gb_pipeline = {
    "model__n_estimators": [100, 200, 300],
    "model__learning_rate": [0.03, 0.05, 0.1],
    "model__max_depth": [2, 3, 4]
}
```

```python
grid_gb_pipeline = GridSearchCV(
    estimator=gb_pipeline,
    param_grid=param_grid_gb_pipeline,
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)
```

```python
grid_gb_pipeline.fit(X_train, y_train_log)
```

```python
print("Best Parameters:", grid_gb_pipeline.best_params_)
print("Best CV RMSLE:", -grid_gb_pipeline.best_score_)
```

```python
best_gb_pipeline = grid_gb_pipeline.best_estimator_

y_pred_best_gb_pipeline = best_gb_pipeline.predict(X_test)

rmsle_best_gb_pipeline = np.sqrt(
    np.mean((y_test_log - y_pred_best_gb_pipeline) ** 2)
)

print("Tuned Pipeline GB RMSLE:", rmsle_best_gb_pipeline)
```

```python
cv_scores_tuned_gb = cross_val_score(
    best_gb_pipeline,
    X_train,
    y_train_log,
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

cv_rmsle_tuned_gb = -cv_scores_tuned_gb

print("Tuned GB Fold RMSLE:", cv_rmsle_tuned_gb)
print("Tuned GB Mean RMSLE:", cv_rmsle_tuned_gb.mean())
print("Tuned GB Std RMSLE:", cv_rmsle_tuned_gb.std())
```
